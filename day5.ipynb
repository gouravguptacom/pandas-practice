```python
import pandas as pd
```

```python
data = {
    "name": ["Amit", "Riya", "John", "Priya", "Rahul"],
    "age": [23, 25, 21, 24, 22],
    "city": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai"],
    "salary": [50000, 60000, 45000, 52000, 58000]
}
```

```python
df = pd.DataFrame(data)
```

Task 1:

Find average salary per city

```python
df
```

```python
df.groupby('city')["salary"].mean()
```

Task 2:

Find:

- max salary per city
- min salary per city

```python
df.groupby('city')['salary'].max(), df.groupby('city').min()
```

Task 3:

Find total salary paid per city

```python
df
```

```python
df.groupby("city")["salary"].sum()
```

Task 4 (Important ⚡):

Get multiple aggregations:

👉 For each city:

- avg salary
- max salary
- min salary

```python
df.groupby("city")["salary"].agg(["mean", "max", "min"])
```

Task 5 (Advanced 🔥):

Find city with highest average salary

```python
c = df.groupby('city')["salary"].mean()
c.idxmax(), c.max()
```
