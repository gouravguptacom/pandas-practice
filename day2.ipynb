```python
import pandas as pd
```

```python
data = {
    "name": ["Amit", "Riya", "John", "Priya", "Rahul"],
    "age": [23, 25, 21, 24, 22],
    "city": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai"],
    "salary": [50000, 60000, 45000, 52000, 58000]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df["name"], df["name"][0] # indexing will work with series only
```

```python
df[["name"]] # this is data fram
df["name"] # this is series
```

```python
df.loc[0, 'name'], df["name"][0]

# df["name"][0] = Not recommended
# df["name"][0] = because it can break with index changes

# Always prefer = df.loc[0, "name"]
```

```python
df.iloc[0, 0]
```

```python
# 🔹 .loc → uses labels (names)
# 🔹 .iloc → uses index (numbers)
```

Task 1:

Get first row using .iloc

```python
df.iloc[[0]]
```

Task 2:

Get first 2 rows and first 2 columns using .iloc

```python
df.iloc[:2, :2]
```

Task 3:

Get row with index 3 using .loc

```python
df.loc[[3]]
```

Task 4:

Get "name" and "city" columns for rows 1 to 3 using .loc

```python
df.loc[1:3, ["name", "city"]]
```

Task 5 (Important ⚡):

Get all rows where:

- age < 24
    Return:
- all columns

(use .loc)

```python
df.loc[df["age"] < 24]
```

| Feature      | `.loc`    | `.iloc`   |
| ------------ | --------- | --------- |
| Uses         | Labels    | Index     |
| Range        | Inclusive | Exclusive |
| Safer for ML | ✅         | ⚠️        |
