```python
import pandas as pd
```

```python
data = {
    "name": ["Amit", "Riya", "John", "Priya", "Rahul"],
    "age": [23, 25, 21, 24, 22],
    "city": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai"],
    "salary": [50000, 60000, 45000, 52000, 58000]
}
```

```python
df = pd.DataFrame(data)
```

Task 1:

Sort data by salary (descending)

```python
df.sort_values(["salary"], ascending=False)
```

Task 2:

Find:
- average salary
- max salary
- min salary

```python
df["salary"].mean(), df["salary"].max(), df["salary"].min()
```

Task 3:

Count how many people are from each city/location

```python
df.groupby("city").count()
```

```python
df['city'].value_counts()
df.groupby('city').size()
df['city'].value_counts().reset_index()
df['city'].value_counts().reset_index(name='count').rename(columns={'index': 'city'})
df['city'].value_counts(ascending=True)
df['city'].value_counts(dropna=False)
```

Task 4:

Find person with highest salary

```python
df.loc[df["salary"] == df["salary"].max()]
# df.nlargest(1, "salary")
```

Task 5 (Important ⚡):

Get all people whose salary is above average

```python
df.loc[df["salary"] > df["salary"].mean()]
```
