```python
import pandas as pd
```

```python
data = {
    "name": ["Amit", "Riya", "John", "Priya", "Rahul"],
    "age": [23, 25, 21, 24, 22],
    "city": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai"],
    "salary": [50000, 60000, 45000, 52000, 58000]
}
```

```python
df = pd.DataFrame(data)
```

Task 1:

Create a new column:
👉 "bonus" = salary * 10%

```python
df["bonus"] = df["salary"] * .10
```

```python
df
```

Task 2:

Create a new column:
👉 "total_income" = salary + bonus

```python
df["total_income"] = df["salary"] + df["bonus"]
```

```python
df
```

Task 3:

Increase salary of people from "Delhi" by 10%

```python
df[df["city"] == 'Delhi']
```

```python
# df["salary"] = df[df["city"] == 'Delhi']['salary'] * .10 = NOT WORKING

df['salary'] = df['salary'].astype(float) # required since it is going to mulitply by float number
df.loc[df['city'] == 'Delhi', 'salary'] *= 1.10
```

```python
df
```

Task 4 (Important ⚡):

Rename column "city" → "location"

```python
df.rename(columns={ "city": "location" }, inplace=True)
```

```python
df
```

Task 5 (Advanced 🔥):

Convert all names to uppercase

```python
df["name"] = df["name"].str.upper()
```
