```python
import pandas as pd
```

```python
data = {
    "name": ["Amit", "Riya", "John", "Priya", "Rahul"],
    "age": [23, 25, 21, 24, 22],
    "city": ["Delhi", "Mumbai", "Delhi", "Kolkata", "Mumbai"],
    "salary": [50000, 60000, 45000, 52000, 58000]
}
```

```python
df = pd.DataFrame(data)
```

Task 5 (Important ⚡):

Get name and salary of people whose salary > 50000

```python
# df[df["salary"] > 50000][["name", "salary"]]
df.loc[df["salary"] > 50000, ["name", "salary"]]
```
